## Rectangular data

Many datasets used in Data Science are tabular, like a spreadsheet or database table. A rectangular dataset has rows and columns. If it has `n` rows and `m` columns, its dimensions are `n × m`.

* A **column** in the table is commonly referred to as a **variable**.
    * A variable may also be called an attribute, input, predictor, or feature.
* A **row** in the table is commonly referred to as a **record**.
    * A record may also be called a case, example, instance, observation, pattern, or sample.


A rectangular dataset can therefore be understood as a two-dimensional structure: rows identify observations and columns describe characteristics of those observations.

What are independent and dependent variables?  

<details>
<p>
    
- An **independent variable** is used to explain or predict another variable. In an experiment, it may be changed or controlled to observe its effect.  
    
- A **dependent variable** is the result being measured, explained, or predicted.
    
</p>
</details>

In Data Science, we often investigate how independent variables relate to or help predict a dependent variable.

What other names are given to these variables?

<details>
<p>

An **independent variable** may also be called a predictor, explanatory variable, feature, input, covariate, or risk factor.
    
A **dependent variable** may also be called a response, outcome, target, label, or output.


    
</p>
</details>

What is Pandas?

Pandas is a Python library for organizing, cleaning, transforming, and analyzing tabular data. Its two main data structures are:

- **Series**: a one-dimensional sequence of values with labels. A single DataFrame column is a Series.
- **DataFrame**: a two-dimensional table with labeled rows and columns. Different columns may contain different data types.

In this notebook, Pandas will be used to create DataFrames, read and write CSV files, inspect datasets, select rows and columns, and filter records using logical conditions. You can inspect any Pandas object or function inside Jupyter with `help(...)` or by placing `?` after its name, for example `pd.DataFrame?`.

#### Select rows that meet logical criteria
- select all students with `height` of **170cm or higher**

````python
students_df[students_df['height'] >= 170]
````

````python
filter_170 = students_df['height'] >= 170

students_df[filter_170]
````

#### Select rows based on multiple conditions
- select all `male` students with `height` of **172cm or higher**

````python
students_df[(students_df['gender'] == 'male') & (students_df['height'] >= 172)]
````

````python
filter_male = students_df['gender'] == 'male'
filter_172 = students_df['height'] >= 172

students_df[filter_male & filter_172]
````

- select all `female` students with `height` of **172cm or higher**

````python
filter_female = students_df['gender'] == 'female'
filter_172 = students_df['height'] >= 172

students_df[filter_female & filter_172]
````


### Subset of both rows and columns
- select the rows where the `shoe` size is **greater than 40** and **only** the columns `name` and `shoe`

### Summing Up

* When selecting subsets of data, square brackets `[]` are used.
* Inside these brackets, you can use a single column/row label, a list of column/row labels, a slice of labels, a conditional expression or a column.
* Select specific rows and/or columns using **loc** when using the row and column names.
* Select specific rows and/or columns using **iloc** when using the positions in the table.
* You can assign new values to a selection based on **loc/iloc**.

#### Quick reference

```python
df["column"]                              # one column
df[["column_a", "column_b"]]            # multiple columns
df[df["column"] >= value]                 # rows matching one condition
df[(condition_a) & (condition_b)]          # both conditions must be true
df[(condition_a) | (condition_b)]          # at least one condition must be true
df.loc[row_condition, ["column_a", "column_b"]]  # rows and named columns
df.iloc[row_positions, column_positions]   # rows and columns by position
```

Use parentheses around every condition combined with `&` or `|`.
